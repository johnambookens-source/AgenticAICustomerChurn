## 1. What is an AI Solutions Architect?

An **AI Solutions Architect** designs the complete system required to turn AI capabilities into a reliable production solution.

The architect sits between:

**Business requirements**
↓
**AI capabilities**
↓
**Software architecture**
↓
**Data & infrastructure**
↓
**Security & governance**
↓
**Production operations**

The architect is therefore not simply responsible for selecting an AI model.

### The key question is:

> How do we build an AI-enabled system that reliably solves the intended business problem under real-world constraints?

---

### A useful mental model

A prototype asks:

> "Can the model do this?"

A production architecture asks:

> "Can the entire system do this reliably, safely, economically, repeatedly, and at scale?"

### AI-Powered Loan Operations Assistant

Our organization processes thousands of loan applications every day.

Loan-operations employees currently spend significant time:

1. Reading application information
2. Checking internal policies and SOPs
3. Identifying missing documents
4. Understanding previous customer interactions
5. Determining whether an application requires escalation
6. Updating internal systems
7. Recording the reasoning and actions taken

The organization wants an AI assistant to support this workflow.

---

### Important constraint

The AI is **not automatically authorized to make every decision**.

Some activities can be automated.

Some activities require:

> AI recommendation → Human review → Final action

- Agents
- Tools

# 4. The Five Questions Every AI Architect Should Ask

Before selecting a model or designing an agent, ask:

---

## Question 1 — What business problem are we solving?

Not:

> "We want an AI agent."

Instead:

> "We want to reduce the time loan-operations employees spend investigating applications while maintaining decision quality and regulatory controls."

---

## Question 2 — Who will use the system?

Possible users:

- Loan operations analyst
- Senior reviewer
- Risk/compliance team
- Operations manager

Different users may require:

- Different permissions
- Different information
- Different workflows
- Different levels of automation

---

## Question 3 — What decisions or actions will the system support?

Examples:

### Information

"Show me the relevant policy."

### Analysis

"Which documents are missing?"

### Recommendation

"Does this application require escalation?"

### Action

"Create a follow-up task."

### High-risk action

"Approve the loan."

These are **not equivalent**.

The architecture must reflect the risk associated with each action.

## Question 4 — What constraints must the system operate within?

Production systems operate under constraints.

For our loan-operations assistant:

### Data constraints

- Customer information is sensitive.
- Users should only access authorized applications.
- Internal policy documents must remain controlled.

### Performance constraints

- Users expect responses within a few seconds for common tasks.
- Long-running investigations may require asynchronous processing.

### Reliability constraints

- The system should remain usable even when one dependency fails.
- Tool failures must not silently produce incorrect actions.

### Security constraints

- Authentication is required.
- Authorization must be enforced.
- Tool permissions must be restricted.

### Cost constraints

- Expensive models should not be used for every task.
- Repeated information retrieval should be optimized.

### Governance constraints

- Important AI outputs must be traceable.
- Actions must be auditable.
- High-risk decisions may require human approval.

## Question 5 — What could go wrong in production?

This is where architecture thinking becomes different from prototype thinking.

Imagine the following scenarios:

| Failure | Possible consequence |
|---|---|
| LLM produces an incorrect explanation | Analyst makes a poor decision |
| Retrieval returns outdated policy | Incorrect recommendation |
| Tool receives incorrect parameters | Wrong system update |
| Authorization is missing | Unauthorized data access |
| Tool becomes unavailable | Workflow interruption |
| Model response becomes very slow | Poor user experience |
| Agent enters a loop | Cost explosion |
| Context becomes too large | Increased cost / degraded output |
| AI performs an action without approval | Governance violation |

The architect's job is not to assume these failures won't happen.

The architect asks:

> **How do we design the system assuming components can fail?**

# 13. Functional vs Non-Functional Requirements

A useful distinction:

## Functional requirements

Describe **what the system does**.

Examples:

- Retrieve policies
- Analyze applications
- Identify missing documents
- Create follow-up tasks
- Escalate cases

---

## Non-functional requirements

Describe **how the system must behave**.

Examples:

- Response time
- Availability
- Scalability
- Security
- Cost
- Auditability
- Maintainability

---

### Why this matters

Two systems can provide exactly the same functionality but have completely different architectures because their non-functional requirements differ.

For example:

### System A

10 users, low-risk internal experimentation.

### System B

500 users, sensitive financial data, audit requirements, high-risk decisions.

Both may use an LLM.

But their production architectures should not be assumed to be identical.

## Tools

A tool is a controlled interface through which the AI system interacts with external information or actions.

Examples:

### Read tools

- get_application()
- get_document_status()
- search_policy()

### Action tools

- create_followup_task()
- update_application()
- notify_reviewer()

---

## Why tools matter

Without tools, the model only knows what is provided in its context.

With tools, the system can interact with current enterprise information.

```text
Model
  │
  │ request

  ▼

  │ result
  ▼
Model / Orchestrator

### Selecting a Foundation Model

Model selection should begin with requirements.

Consider these dimensions:

| Dimension | Question |
|---|---|
| Capability | Can it perform the required task? |
| Reasoning | Can it handle task complexity? |
| Context | Can it process the required context? |
| Latency | Is response time acceptable? |
| Cost | Is the cost acceptable at scale? |
| Reliability | Is behaviour sufficiently consistent? |
| Tool use | Can it interact with required tools? |
| Structured output | Can it reliably produce required formats? |
| Availability | Is the service available in the target environment? |
| Governance | Does it satisfy organizational requirements? |

---

## Important

There is rarely a single "best model."

Instead:

> **Select the model that satisfies the requirements at an acceptable cost and operational complexity.**
