```python
@dataclass
class AgentState:
    user_goal: str
    execution_history: List[Dict[str, Any]] = field(default_factory=list)
    observations: List[str] = field(default_factory=list)
    reasoning_outputs: List[str] = field(default_factory=list)
    tool_responses: List[Dict[str, Any]] = field(default_factory=list)
    final_outcome: Optional[str] = None
    iteration: int = 0
    status: str = "RUNNING"
```

```python
def display_state(state: AgentState):
    print("=" * 70)
    print(f"ITERATION : {state.iteration}")
    print(f"STATUS    : {state.status}")
    print(f"GOAL      : {state.user_goal}")

    print("\nOBSERVATIONS:")
    for item in state.observations:
        print(" -", item)

    print("\nHISTORY:")
    for item in state.execution_history:
        print(" -", item)

    print("\nFINAL OUTCOME:")
    print(state.final_outcome)

    print("=" * 70)


display_state(state)
```

```python
def build_reasoning_prompt(state: AgentState) -> str:
    return f"""
You are the reasoning component of a retail customer intelligence agent.

USER GOAL:
{state.user_goal}

CURRENT ITERATION:
{state.iteration}

OBSERVATIONS:
{json.dumps(state.observations, indent=2)}

EXECUTION HISTORY:
{json.dumps(state.execution_history, indent=2)}

CUSTOMER DATA:
{json.dumps(dataset_summary, indent=2)}

Your task:

1. Analyse the current state.
2. Determine what should happen next.
3. Decide whether the business goal has been achieved.

Respond with exactly this structure:

NEXT_ACTION: <describe the next action>
GOAL_COMPLETE: <YES or NO>
OUTCOME: <short explanation>
"""
```

```python
def reason(state: AgentState) -> str:

    prompt = build_reasoning_prompt(state)

    response = client.responses.create(
        model="gpt-4.1-mini",
        input=prompt
    )

    reasoning = response.output_text

    state.reasoning_outputs.append(reasoning)

    return reasoning


reasoning = reason(state)

print(reasoning)
```

```python
def parse_decision(reasoning: str) -> Dict[str, str]:

    decision = {
        "next_action": "",
        "goal_complete": "",
        "outcome": ""
    }

    for line in reasoning.splitlines():

        if line.startswith("NEXT_ACTION:"):
            decision["next_action"] = line.replace(
                "NEXT_ACTION:", ""
            ).strip()

        elif line.startswith("GOAL_COMPLETE:"):
            decision["goal_complete"] = line.replace(
                "GOAL_COMPLETE:", ""
            ).strip().upper()

        elif line.startswith("OUTCOME:"):
            decision["outcome"] = line.replace(
                "OUTCOME:", ""
            ).strip()

    return decision


decision = parse_decision(reasoning)

decision
```

```python
def act(state: AgentState, decision: Dict[str, str]):

    state.execution_history.append({
        "iteration": state.iteration,
        "action": decision["next_action"],
        "result": decision["outcome"]
    })

    state.observations.append(
        f"Agent decided: {decision['next_action']}"
    )

    if decision["goal_complete"] == "YES":
        state.final_outcome = decision["outcome"]
        state.status = "COMPLETED"


act(state, decision)

display_state(state)
```

```python
def run_agent_once(state: AgentState):

    print(f"\n--- ITERATION {state.iteration} ---")

    # OBSERVE
    print("1. OBSERVE")

    # REASON
    print("2. REASON")
    reasoning = reason(state)

    # DECIDE
    print("3. DECIDE")
    decision = parse_decision(reasoning)
    print("Decision:", decision)

    # ACT
    print("4. ACT")
    act(state, decision)

    # UPDATE
    print("5. UPDATE STATE")
    state.iteration += 1

    return state
```

```python
def run_agent(state: AgentState, max_iterations: int = 3):

    while (
        state.status == "RUNNING"
        and state.iteration < max_iterations
    ):

        state = run_agent_once(state)

        print("\nCurrent status:", state.status)

    return state
```

```python
def get_customer(customer_id: str) -> Dict[str, Any]: #Retrieval Tool

    matches = customers[
        customers["customer_id"] == customer_id
    ]

    if matches.empty:
        raise ValueError(
            f"Customer '{customer_id}' was not found."
        )

    return matches.iloc[0].to_dict()
```

```python
def get_customer_validated(customer_id: str) -> Dict[str, Any]:

    if not isinstance(customer_id, str):
        raise TypeError(
            "customer_id must be a string."
        )

    if not customer_id.strip():
        raise ValueError(
            "customer_id cannot be empty."
        )

    customer_id = customer_id.strip().upper()

    customer = get_customer(customer_id)

    required_fields = [
        "customer_id",
        "name",
        "monthly_spend",
        "orders_last_90d",
        "days_since_last_order",
        "support_tickets",
        "discount_usage_pct"
    ]

    missing = [
        field for field in required_fields
        if field not in customer
    ]

    if missing:
        raise ValueError(
            f"Tool response missing fields: {missing}"
        )

    return customer
```

```python
def calculate_risk_score(customer: Dict[str, Any]) -> Dict[str, Any]:

    score = 0
    reasons = []

    if customer["days_since_last_order"] > 60:
        score += 40
        reasons.append("Long inactivity period")

    elif customer["days_since_last_order"] > 30:
        score += 25
        reasons.append("Increasing inactivity")

    if customer["orders_last_90d"] <= 2:
        score += 25
        reasons.append("Low recent order frequency")

    if customer["support_tickets"] >= 5:
        score += 20
        reasons.append("High support interaction")

    if customer["discount_usage_pct"] >= 80:
        score += 15
        reasons.append("High discount dependency")

    score = min(score, 100)

    if score >= 60:
        risk_level = "HIGH"
    elif score >= 30:
        risk_level = "MEDIUM"
    else:
        risk_level = "LOW"

    return {
        "customer_id": customer["customer_id"],
        "risk_score": score,
        "risk_level": risk_level,
        "reasons": reasons
    }
```

```python
def choose_tool(state: AgentState) -> Dict[str, Any]:

    prompt = f"""
You are the decision-making component of a customer intelligence agent.

USER GOAL:
{state.user_goal}

CURRENT STATE:
{json.dumps(state.observations, indent=2)}

AVAILABLE TOOLS:

1. get_customer
   Purpose: Retrieve customer information.
   Input: customer_id

2. calculate_risk
   Purpose: Calculate churn risk.
   Input: customer object

Choose the next tool required.

Return ONLY valid JSON:

{{
    "tool": "<tool name>",
    "arguments": {{...}}
}}
"""

    response = client.responses.create(
        model="gpt-4.1-mini",
        input=prompt
    )

    return json.loads(response.output_text)
```

```python
TOOL_SCHEMAS = {

    "get_customer": {
        "required": ["customer_id"],#mandatory
        "types": {
            "customer_id": str
        }
    },

    "calculate_risk": {
        "required": ["customer"],
        "types": {
            "customer": dict
        }
    }
}
```

```python
def validate_tool_call(
    tool_name: str,
    arguments: Dict[str, Any]
):

    if tool_name not in TOOL_SCHEMAS:
        raise ValueError(
            f"Unknown tool: {tool_name}"
        )

    schema = TOOL_SCHEMAS[tool_name]

    for required_field in schema["required"]:
        if required_field not in arguments:
            raise ValueError(
                f"Missing required parameter: "
                f"{required_field}"
            )

    for field, expected_type in schema["types"].items():

        if field in arguments:
            if not isinstance(
                arguments[field],
                expected_type
            ):
                raise TypeError(
                    f"{field} must be "
                    f"{expected_type.__name__}"
                )

    return True
```

```python
def execute_tool(
    tool_name: str,
    arguments: Dict[str, Any]
):

    validate_tool_call(
        tool_name,
        arguments
    )

    if tool_name not in TOOLS:
        raise ValueError(
            f"Tool '{tool_name}' is not registered."
        )

    tool_function = TOOLS[tool_name]

    return tool_function(**arguments)
```

```python
result = execute_tool(
    "get_customer",
    {
        "customer_id": "C002"
    }
)

print(json.dumps(result, indent=2))
```

```python
def execute_tool_with_retry(
    tool_name: str,
    arguments: Dict[str, Any],
    max_retries: int = 2
):

    last_error = None

    for attempt in range(1, max_retries + 1):

        try:

            result = execute_tool(
                tool_name,
                arguments
            )

            return {
                "success": True,
                "attempt": attempt,
                "result": result
            }

        except Exception as e:

            last_error = str(e)

            print(
                f"Attempt {attempt} failed: "
                f"{last_error}"
            )

    return {
        "success": False,
        "attempt": max_retries,
        "error": last_error,
        "fallback": (
            "Unable to execute the requested tool. "
            "Return a safe explanation instead."
        )
    }
```

```python
def customer_analysis_skill(customer_id: str):

    # Step 1 — Retrieve customer
    customer_result = execute_tool_with_retry(
        "get_customer",
        {"customer_id": customer_id}
    )

    if not customer_result["success"]:
        return {
            "success": False,
            "stage": "customer_retrieval",
            "error": customer_result["error"]
        }

    customer = customer_result["result"]

    # Step 2 — Calculate risk
    risk_result = execute_tool_with_retry(
        "calculate_risk",
        {"customer": customer}
    )

    if not risk_result["success"]:
        return {
            "success": False,
            "stage": "risk_calculation",
            "error": risk_result["error"]
        }

    risk = risk_result["result"]

    return {
        "success": True,
        "customer": customer,
        "risk": risk
    }
```

```python
def explain_customer_risk(
    customer: Dict[str, Any],
    risk: Dict[str, Any]
) -> str:

    prompt = f"""
You are a retail customer analytics expert.

CUSTOMER:
{json.dumps(customer, indent=2)}

RISK ASSESSMENT:
{json.dumps(risk, indent=2)}

Explain:

1. Why this customer has this risk level.
2. Which factors are most important.
3. What the business should pay attention to.

Keep the explanation concise and business-focused.
"""

    response = client.responses.create(
        model="gpt-4.1-mini",
        input=prompt
    )

    return response.output_text
```

```python
def intervention_recommendation_skill(
    customer: Dict[str, Any],
    risk: Dict[str, Any]
) -> Dict[str, Any]:

    risk_level = risk["risk_level"]

    if risk_level == "HIGH":

        action = "Immediate retention intervention"
        priority = "P1"

    elif risk_level == "MEDIUM":

        action = "Targeted engagement campaign"
        priority = "P2"

    else:

        action = "Continue regular engagement"
        priority = "P3"

    return {
        "customer_id": customer["customer_id"],
        "risk_level": risk_level,
        "recommended_action": action,
        "priority": priority
    }
```

```python
def churn_assessment_skill(customer_id: str):

    # STEP 1 — Customer analysis
    analysis = customer_analysis_skill(customer_id)

    if not analysis["success"]:
        return {
            "success": False,
            "stage": "analysis",
            "error": analysis["error"]
        }

    # STEP 2 — Explain risk
    explanation = explain_customer_risk(
        analysis["customer"],
        analysis["risk"]
    )

    # STEP 3 — Generate recommendation
    recommendation = intervention_recommendation_skill(
        analysis["customer"],
        analysis["risk"]
    )

    return {
        "success": True,
        "customer": analysis["customer"],
        "risk": analysis["risk"],
        "explanation": explanation,
        "recommendation": recommendation
    }
```

```python
@dataclass
class TraceEvent:
    iteration: int
    event_type: str
    details: Dict[str, Any]


@dataclass
class AgentTrace:
    events: List[TraceEvent] = field(
        default_factory=list
    )

    def log(
        self,
        iteration: int,
        event_type: str,
        details: Dict[str, Any]
    ):

        self.events.append(
            TraceEvent(
                iteration=iteration,
                event_type=event_type,
                details=details
            )
        )

    def show(self):

        for event in self.events:

            print(
                f"[Iteration {event.iteration}] "
                f"{event.event_type}"
            )

            print(
                json.dumps(
                    event.details,
                    indent=2
                )
            )

            print("-" * 60)
```

```python
def check_goal_completion(
    state: AgentState
) -> bool:

    if not state.final_outcome:
        return False

    outcome_text = state.final_outcome.lower()

    required_terms = [
        "risk",
        "recommend"
    ]

    return all(
        term in outcome_text
        for term in required_terms
    )
```

```python
def should_terminate(
    state: AgentState,
    max_steps: int = 6
) -> bool:

    # Condition 1: explicit completion
    if state.status == "COMPLETED":
        return True

    # Condition 2: goal achieved
    if check_goal_completion(state):
        state.status = "COMPLETED"
        return True

    # Condition 3: step limit
    if state.iteration >= max_steps:
        state.status = "MAX_STEPS_REACHED"
        return True

    # Condition 4: unrecoverable failure
    if state.status == "FAILED":
        return True

    return False
```

```python
def generate_business_response(
    assessment: Dict[str, Any]
) -> str:

    if not assessment["success"]:
        return (
            "The customer assessment could not be completed. "
            f"Reason: {assessment['error']}"
        )

    customer = assessment["customer"]
    risk = assessment["risk"]
    recommendation = assessment["recommendation"]

    return f"""
Customer: {customer['name']} ({customer['customer_id']})

Churn Risk: {risk['risk_level']}
Risk Score: {risk['risk_score']}/100

Key Risk Factors:
- {"; ".join(risk['reasons'])}

Recommendation:
{recommendation['recommended_action']}

Priority:
{recommendation['priority']}
"""
```

```python
def reflect_on_result(
    user_goal: str,
    result: str
) -> Dict[str, Any]:

    prompt = f"""
You are a quality-control component for an AI agent.

USER GOAL:
{user_goal}

AGENT RESULT:
{result}

Evaluate whether the result adequately satisfies the goal.

Return ONLY JSON:

{{
    "approved": true or false,
    "issues": ["issue 1", "issue 2"],
    "correction": "what should be improved"
}}
"""

    response = client.responses.create(
        model="gpt-4.1-mini",
        input=prompt
    )

    return json.loads(response.output_text)
```

```python
def correct_result(
    user_goal: str,
    result: str,
    reflection: Dict[str, Any]
) -> str:

    prompt = f"""
Improve the following agent result.

USER GOAL:
{user_goal}

CURRENT RESULT:
{result}

QUALITY REVIEW:
{json.dumps(reflection, indent=2)}

Produce a corrected business response.
"""

    response = client.responses.create(
        model="gpt-4.1-mini",
        input=prompt
    )

    return response.output_text
```

```python
def run_agent(user_goal: str, max_steps: int = 6):

    # ============================================================
    # 1. INITIALIZE AGENT STATE
    # ============================================================

    state = AgentState(
        user_goal=user_goal
    )

    trace = AgentTrace()

    trace.log(
        0,
        "AGENT_START",
        {
            "goal": user_goal
        }
    )

    # ============================================================
    # 2. AGENT LOOP
    # ============================================================

    while not should_terminate(
        state,
        max_steps=max_steps
    ):

        iteration = state.iteration

        # --------------------------------------------------------
        # OBSERVE
        # --------------------------------------------------------

        observation = {
            "goal": state.user_goal,
            "history": state.execution_history[-3:],
            "tool_results": state.tool_responses[-3:],
            "observations": state.observations[-3:]
        }

        trace.log(
            iteration,
            "OBSERVE",
            observation
        )

        # --------------------------------------------------------
        # REASON + DECIDE
        # --------------------------------------------------------

        decision = choose_skill(
            state.user_goal
        )

        trace.log(
            iteration,
            "DECIDE",
            decision
        )

        state.execution_history.append({
            "iteration": iteration,
            "action": "SELECT_SKILL",
            "decision": decision
        })

        # --------------------------------------------------------
        # VALIDATE DECISION
        # --------------------------------------------------------

        selected_skill = decision.get("skill")
        skill_arguments = decision.get(
            "arguments",
            {}
        )

        if selected_skill not in SKILLS:

            trace.log(
                iteration,
                "INVALID_SKILL",
                {
                    "skill": selected_skill
                }
            )

            state.status = "FAILED"

            break

        # --------------------------------------------------------
        # ACT — EXECUTE SKILL
        # --------------------------------------------------------

        trace.log(
            iteration,
            "SKILL_START",
            {
                "skill": selected_skill,
                "arguments": skill_arguments
            }
        )

        try:

            skill_result = SKILLS[selected_skill](
                **skill_arguments
            )

        except Exception as e:

            trace.log(
                iteration,
                "SKILL_ERROR",
                {
                    "skill": selected_skill,
                    "error": str(e)
                }
            )

            state.status = "FAILED"

            break

        # --------------------------------------------------------
        # UPDATE STATE
        # --------------------------------------------------------

        state.tool_responses.append(
            skill_result
        )

        state.observations.append(
            f"Skill '{selected_skill}' completed."
        )

        state.execution_history.append({
            "iteration": iteration,
            "action": "EXECUTE_SKILL",
            "skill": selected_skill,
            "result": skill_result
        })

        trace.log(
            iteration,
            "SKILL_RESULT",
            {
                "skill": selected_skill,
                "success": skill_result.get(
                    "success",
                    False
                )
            }
        )

        # --------------------------------------------------------
        # GENERATE PROVISIONAL ANSWER
        # --------------------------------------------------------

        if skill_result.get("success"):

            provisional_answer = (
                generate_business_response(
                    skill_result
                )
            )

            state.final_outcome = (
                provisional_answer
            )

        else:

            state.final_outcome = (
                f"Skill failed: "
                f"{skill_result.get('error')}"
            )

        # --------------------------------------------------------
        # REFLECTION / QUALITY CHECK
        # --------------------------------------------------------

        reflection = reflect_on_result(
            user_goal=state.user_goal,
            result=state.final_outcome
        )

        trace.log(
            iteration,
            "REFLECTION",
            reflection
        )

        # --------------------------------------------------------
        # SELF-CORRECTION
        # --------------------------------------------------------

        if not reflection["approved"]:

            trace.log(
                iteration,
                "SELF_CORRECTION",
                {
                    "issues": reflection["issues"]
                }
            )

            corrected_answer = correct_result(
                user_goal=state.user_goal,
                result=state.final_outcome,
                reflection=reflection
            )

            state.final_outcome = (
                corrected_answer
            )

            state.observations.append(
                "Agent performed self-correction."
            )

        # --------------------------------------------------------
        # GOAL COMPLETION CHECK
        # --------------------------------------------------------

        if check_goal_completion(state):

            state.status = "COMPLETED"

            trace.log(
                iteration,
                "GOAL_COMPLETED",
                {
                    "outcome": state.final_outcome
                }
            )

        else:

            state.iteration += 1

            trace.log(
                iteration,
                "GOAL_NOT_COMPLETED",
                {
                    "next_iteration":
                        state.iteration
                }
            )

    # ============================================================
    # 3. FINAL TERMINATION
    # ============================================================

    if state.status != "COMPLETED":

        if state.iteration >= max_steps:

            state.status = "MAX_STEPS_REACHED"

            trace.log(
                state.iteration,
                "TERMINATED",
                {
                    "reason": "Maximum step limit reached."
                }
            )

        elif state.status != "FAILED":

            state.status = "TERMINATED"

            trace.log(
                state.iteration,
                "TERMINATED",
                {
                    "reason":
                        "Agent termination condition reached."
                }
            )

    # ============================================================
    # 4. FINAL TRACE
    # ============================================================

    trace.log(
        state.iteration,
        "AGENT_END",
        {
            "status": state.status
        }
    )

    return {
        "status": state.status,
        "answer": state.final_outcome,
        "state": state,
        "trace": trace
    }
```

```python
result = run_agent(
    """
    Assess customer C002 for churn risk,
    explain the major risk factors,
    and recommend the most appropriate
    retention action.
    """
)
```
